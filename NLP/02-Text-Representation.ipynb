### 1. What is BoW? (Theory in Simple Words)

- Imagine you have a bag 👜 where you put all the words from a text, without caring about order or grammar.

The only thing that matters → how many times each word appears.
    

👉 That’s why it’s called Bag of Words:

- Words = tokens

- Bag = collection where order doesn’t matter

```python
from sklearn.feature_extraction.text import CountVectorizer

docs = ["I love NLP", "NLP loves Python", "Python loves AI"]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(docs)

print("Vocabulary:", vectorizer.vocabulary_)
print("BOW matrix:\n", X.toarray())
```

## 2. TF-IDF (Term Frequency – Inverse Document Frequency)

### 1️. What is TF-IDF? (Simplest Explanation)

**BoW problem:**
- In Bag of Words (BoW), all words are treated equally.  
- Example: In 1000 documents, the word *“the”* appears in almost every document → but BoW still counts it as important.  

**TF-IDF solution:**
- It reduces the weight of **common words** (like *“the”*, *“is”*, *“and”*)  
- It increases the weight of **rare but important words** (like *“Python”*, *“machine learning”*, *“spam”*).  

---

### 2️. Formula (Simplified)

**IDF (Inverse Document Frequency)** → How unique the word is across documents.  

$$
IDF = \log \left( \frac{\text{total documents}}{\text{documents containing the word}} \right)
$$
---

**TF-IDF = TF × IDF**

**Meaning:** A word gets a **high score** if:
- It appears many times in a document (**high TF**)  
- But appears rarely across other documents (**high IDF**)  


```python
from sklearn.feature_extraction.text import TfidfVectorizer

docs = ["I love NLP", 
        "NLP loves Python", 
        "Python loves AI"]

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(docs)

print("Vocabulary:", vectorizer.vocabulary_)
print("TF-IDF matrix:\n", X.toarray())
```
